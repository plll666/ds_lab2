Загружаем файл, исследуем листы, структуру первых 15 строк, проверяем объединение ячеек и выводим последние 5 строк

```python
import openpyxl
import os

filename = list(uploaded.keys())[0]
print(f"Загружен файл: {filename}")
wb = openpyxl.load_workbook(filename, data_only=True)


print("листы в файле:")
for i, name in enumerate(wb.sheetnames):
    ws = wb[name]
    visibility = "скрыт" if ws.sheet_state == 'скрыт' else "видим"
    print(f"  [{i}] '{name}' — {visibility}")

for sheet_name in wb.sheetnames:
    ws = wb[sheet_name]
    print(f"лист: '{sheet_name}' | Состояние: {ws.sheet_state}")
    print(f"Размер: {ws.max_row} строк x {ws.max_column} столбцов")

    print(f"  Первые 15 строк:")
    for row_idx, row in enumerate(ws.iter_rows(min_row=1, max_row=15, values_only=False), 1):
        row_data = []
        for cell in row:
            if cell.value is not None:
                row_data.append(f"{cell.coordinate}={cell.value}({cell.data_type})")
        if row_data:
            print(f"    Строка {row_idx}: {' | '.join(row_data)}")
        else:
            print(f"    Строка {row_idx}: [пусто]")

    if ws.merged_cells.ranges:
        print(f"объедененные ячейки :")
        for merged_range in ws.merged_cells.ranges:
            print(f"    {merged_range}")

    print(f"Последние 5 непустых строк:")
    printed = 0
    for row_idx in range(ws.max_row, 0, -1):
        row = ws[row_idx]
        if any(cell.value is not None for cell in row):
            row_data = []
            for cell in row:
                if cell.value is not None:
                    val = str(cell.value)[:50]
                    row_data.append(f"{cell.coordinate}={val}")
            print(f"Строка {row_idx}: {' | '.join(row_data)}")
            printed += 1
            if printed >= 5:
                break
```

Исправление кодировки кириллицы и опечаток в названиях городов. Каждый лист парсится отдельно, применяется `normalize_city`

```python
import pandas as pd
import numpy as np

def normalize_city(name):
    if pd.isna(name):
        return np.nan

    if isinstance(name, str):
        try:
            name = name.encode('latin1').decode('cp1251')
        except:
            pass

    name = str(name).strip()

    typo_map = {
        'Геленджик': ['Геленджикк', 'Геленджи'],
        'Благовещенск': ['Благовещенс', 'Благовещенскк'],
        'Москва': ['Мосва'],
        'Сочи': ['Сычи', 'Счи', 'Соч']
    }

    for correct, variants in typo_map.items():
        if name in variants or name == correct:
            return correct
    return name

df1 = pd.read_excel('have_fun.xlsx', sheet_name='Ëèñò1')
df2 = pd.read_excel('have_fun.xlsx', sheet_name='Ëèñò2_ñòðîêîâûå_NaN_âûáðîñû', usecols=range(10))
df3 = pd.read_excel('have_fun.xlsx', sheet_name='Ëèñò3_ñìåøàííûå_òèïû')

for df_temp in [df1, df2, df3]:
    df_temp['city'] = df_temp['city'].apply(normalize_city)

print("Результаты после исправления кодировки:")
print(f"Лист 1: {df1['city'].unique()}")
print(f"Лист 2: {df2['city'].unique()}")
print(f"Лист 3: {df3['city'].unique()}")
```

Приведение городов к единому регистру, конвертация числовых колонок, объединение трёх листов в один датафрейм, удаление строк без даты или города.

```python
for df_temp in [df1, df2, df3]:
    df_temp['city'] = df_temp['city'].str.capitalize()

    numeric_cols = ['temperature_2m', 'relative_humidity_2m', 'precipitation', 'rain', 'snowfall', 'wind_speed_10m', 'surface_pressure']
    for col in numeric_cols:
        if col in df_temp.columns:
            df_temp[col] = pd.to_numeric(df_temp[col], errors='coerce')

df_full = pd.concat([df1, df2, df3], ignore_index=True)

df_full = df_full.dropna(subset=['ds', 'city'])

print("Итоговый список городов после унификации:")
print(df_full['city'].unique())
print(f"\nОбщая форма данных: {df_full.shape}")
display(df_full.sample(5))
```

Приведение к регулярному часовому ряду для каждого города, удаление дубликатов, интерполяция пропусков, очистка выбросов температуры за пределами `[-60, 60]`



```python
df_resampled = []

for city in df_full['city'].unique():
    city_df = df_full[df_full['city'] == city].copy()
    city_df = city_df.sort_values('ds')
    city_df = city_df.set_index('ds')

    city_df = city_df[~city_df.index.duplicated(keep='first')]

    city_df = city_df.asfreq('H')
    city_df['city'] = city.capitalize()

    num_cols = city_df.select_dtypes(include=[np.number]).columns
    city_df[num_cols] = city_df[num_cols].interpolate(method='linear')

    if 'temperature_2m' in city_df.columns:
        city_df.loc[(city_df['temperature_2m'] > 60) | (city_df['temperature_2m'] < -60), 'temperature_2m'] = np.nan
        city_df['temperature_2m'] = city_df['temperature_2m'].interpolate(method='linear')

    df_resampled.append(city_df.reset_index())

df_final = pd.concat(df_resampled, ignore_index=True)

print("Статистика после обработки временных рядов:")
print(df_final.describe())
print(f"Итоговое количество пропусков:{df_final.isnull().sum()}")
```

Анализ временного ряда: график всей истории, декомпозиция на тренд/сезонность/остатки `(последние 30 дней, период 24ч)`, тест Дики-Фуллера на стационарность.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller

city_to_analyze = 'Сочи'
city_data = df_final[df_final['city'] == city_to_analyze].set_index('ds')

plt.figure(figsize=(15, 6))
plt.plot(city_data['temperature_2m'], label='Температура')
plt.title(f'Динамика температуры в городе {city_to_analyze}')
plt.legend()
plt.show()

result = seasonal_decompose(city_data['temperature_2m'].tail(24*30), model='additive', period=24)

fig = result.plot()
fig.set_size_inches(15, 10)
plt.show()

print(f"Тест Дики-Фуллера для {city_to_analyze}:")
adf_test = adfuller(city_data['temperature_2m'].dropna())
print(f"ADF Statistic: {adf_test[0]:.4f}")
print(f"p-value: {adf_test[1]:.4f}")
if adf_test[1] < 0.05:
    print("Ряд стационарен")
else:
    print("Ряд нестационарен")
```

Генерация признаков. Включает временные, циклические, лаговые признаки и скользящие статистики. Таргет сдвинут на `1 час` вперёд.

```python
def create_features(df):
    df = df.copy()

    df['hour'] = df['ds'].dt.hour
    df['dayofweek'] = df['ds'].dt.dayofweek
    df['quarter'] = df['ds'].dt.quarter
    df['month'] = df['ds'].dt.month
    df['year'] = df['ds'].dt.year
    df['dayofyear'] = df['ds'].dt.dayofyear
    df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)

    df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
    df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)

    for lag in [1, 2, 3, 6, 12, 24, 48, 168]:
        df[f'lag_{lag}'] = df.groupby('city')['temperature_2m'].shift(lag)

    for window in [6, 12, 24]:
        df[f'rolling_mean_{window}'] = (
            df.groupby('city')['temperature_2m']
            .transform(lambda x: x.shift(1).rolling(window=window, min_periods=1).mean())
        )
        df[f'rolling_std_{window}'] = (
            df.groupby('city')['temperature_2m']
            .transform(lambda x: x.shift(1).rolling(window=window, min_periods=1).std())
        )

    df['target'] = df.groupby('city')['temperature_2m'].shift(-1)

    return df

df_features = create_features(df_final)

features = [
    'hour', 'dayofweek', 'month', 'dayofyear', 'is_weekend',
    'hour_sin', 'hour_cos', 'month_sin', 'month_cos',
    'lag_1', 'lag_2', 'lag_3', 'lag_6', 'lag_12', 'lag_24', 'lag_48', 'lag_168',
    'rolling_mean_6', 'rolling_mean_12', 'rolling_mean_24',
    'rolling_std_6', 'rolling_std_12', 'rolling_std_24'
]
target = 'target'

print(f"Всего признаков: {len(features)}")
print(f"Пропусков в таргете: {df_features['target'].isna().sum()}")
```

Обучение DT, RF, XGBoost с train/val/test (70/15/15). Optuna подбирает гиперпараметры на val, финальная оценка на test. Модели обучаются на train+val.

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error
import optuna
import matplotlib.pyplot as plt

target_city = 'Сочи'
city_df = df_features[df_features['city'] == target_city].dropna().sort_values('ds')

X = city_df[features]
y = city_df[target]

n = len(city_df)
train_idx = int(n * 0.7)
val_idx = int(n * 0.85)

X_train, y_train = X.iloc[:train_idx], y.iloc[:train_idx]
X_val, y_val = X.iloc[train_idx:val_idx], y.iloc[train_idx:val_idx]
X_test, y_test = X.iloc[val_idx:], y.iloc[val_idx:]

print(f"Город: {target_city}")
print(f"Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}")

def objective_dt(trial):
    params = {
        'max_depth': trial.suggest_int('max_depth', 3, 20),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
        'random_state': 42
    }
    model = DecisionTreeRegressor(**params)
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    return mean_absolute_error(y_val, preds)

study_dt = optuna.create_study(direction='minimize')
study_dt.optimize(objective_dt, n_trials=20, show_progress_bar=True)

best_dt = DecisionTreeRegressor(**study_dt.best_params, random_state=42)
best_dt.fit(pd.concat([X_train, X_val]), pd.concat([y_train, y_val]))
dt_preds = best_dt.predict(X_test)

print(f"DT best params: {study_dt.best_params}")
print(f"DT MAE on test: {mean_absolute_error(y_test, dt_preds):.4f}")

def objective_rf(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 200),
        'max_depth': trial.suggest_int('max_depth', 5, 20),
        'random_state': 42
    }
    model = RandomForestRegressor(**params, n_jobs=-1)
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    return mean_absolute_error(y_val, preds)

study_rf = optuna.create_study(direction='minimize')
study_rf.optimize(objective_rf, n_trials=15, show_progress_bar=True)

best_rf = RandomForestRegressor(**study_rf.best_params, random_state=42, n_jobs=-1)
best_rf.fit(pd.concat([X_train, X_val]), pd.concat([y_train, y_val]))
rf_preds = best_rf.predict(X_test)
print(f"RF MAE on test: {mean_absolute_error(y_test, rf_preds):.4f}")

def objective_xgb(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 500),
        'learning_rate': trial.suggest_float('learning_rate', 0.005, 0.1),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'random_state': 42
    }
    model = xgb.XGBRegressor(**params)
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    return mean_absolute_error(y_val, preds)

study_xgb = optuna.create_study(direction='minimize')
study_xgb.optimize(objective_xgb, n_trials=30, show_progress_bar=True)

best_xgb = xgb.XGBRegressor(**study_xgb.best_params, random_state=42)
best_xgb.fit(pd.concat([X_train, X_val]), pd.concat([y_train, y_val]))
xgb_preds = best_xgb.predict(X_test)
print(f"XGB best params: {study_xgb.best_params}")
print(f"XGB MAE on test: {mean_absolute_error(y_test, xgb_preds):.4f}")
```

Визуализация одношаговых прогнозов трёх моделей на последних 120 часах теста

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 7))

last_n = 120
plt.plot(y_test.values[-last_n:], label='Реальная температура', color='black', linewidth=2, marker='o', markersize=4)
plt.plot(dt_preds[-last_n:], label=f'Decision Tree (MAE: {mean_absolute_error(y_test, dt_preds):.2f})', linestyle='--')
plt.plot(rf_preds[-last_n:], label=f'Random Forest (MAE: {mean_absolute_error(y_test, rf_preds):.2f})', linestyle='--')
plt.plot(xgb_preds[-last_n:], label=f'XGBoost (MAE: {mean_absolute_error(y_test, xgb_preds):.2f})', linestyle='-', linewidth=2)

plt.title(f'Сравнение моделей прогнозирования для города {target_city}')
plt.xlabel('Время (последние 120 часов)')
plt.ylabel('Температура (°C)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Сравнение для другого города
Проверка переносимости гиперпараметров с Сочи на Санкт-Петербург. Обучает XGBoost с параметрами Сочи и со своей оптимизацией, сравнивает MAE.

```python
new_target_city = 'Санкт-петербург'
city_df_new = df_features[df_features['city'] == new_target_city].dropna().sort_values('ds')

X_new = city_df_new[features]
y_new = city_df_new[target]

n_new = len(city_df_new)
train_idx_new = int(n_new * 0.7)
val_idx_new = int(n_new * 0.85)

X_tr_new, y_tr_new = X_new.iloc[:train_idx_new], y_new.iloc[:train_idx_new]
X_val_new, y_val_new = X_new.iloc[train_idx_new:val_idx_new], y_new.iloc[train_idx_new:val_idx_new]
X_te_new, y_te_new = X_new.iloc[val_idx_new:], y_new.iloc[val_idx_new:]

print(f"Город: {new_target_city}")
print(f"Train: {len(X_tr_new)} | Val: {len(X_val_new)} | Test: {len(X_te_new)}")

model_xgb_new = xgb.XGBRegressor(**study_xgb.best_params, random_state=42)
model_xgb_new.fit(pd.concat([X_tr_new, X_val_new]), pd.concat([y_tr_new, y_val_new]))
preds_xgb_new = model_xgb_new.predict(X_te_new)

def objective_xgb_new(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 500),
        'learning_rate': trial.suggest_float('learning_rate', 0.005, 0.1),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'random_state': 42
    }
    model = xgb.XGBRegressor(**params)
    model.fit(X_tr_new, y_tr_new)
    preds = model.predict(X_val_new)
    return mean_absolute_error(y_val_new, preds)

study_xgb_new = optuna.create_study(direction='minimize')
study_xgb_new.optimize(objective_xgb_new, n_trials=15, show_progress_bar=True)

model_xgb_opt = xgb.XGBRegressor(**study_xgb_new.best_params, random_state=42)
model_xgb_opt.fit(pd.concat([X_tr_new, X_val_new]), pd.concat([y_tr_new, y_val_new]))
preds_xgb_opt = model_xgb_opt.predict(X_te_new)

mae_transfer = mean_absolute_error(y_te_new, preds_xgb_new)
mae_optimized = mean_absolute_error(y_te_new, preds_xgb_opt)

print(f"Результаты для города: {new_target_city}")
print(f"MAE (параметры от Сочи): {mae_transfer:.4f}")
print(f"MAE (своя оптимизация): {mae_optimized:.4f}")
print(f"Разница: {mae_transfer - mae_optimized:+.4f}°C")

if abs(mae_transfer - mae_optimized) < 0.05:
    print("Вывод: гиперпараметры хорошо переносятся между городами")
else:
    print("Вывод: для каждого города лучше подбирать свои гиперпараметры")

plt.figure(figsize=(15, 7))
plt.plot(y_te_new.values[-168:], label='Реальность', color='black', alpha=0.6)
plt.plot(preds_xgb_opt[-168:], label=f'XGBoost (своя оптимизация, MAE={mae_optimized:.2f})', linewidth=2)
plt.title(f'Прогноз температуры: {new_target_city} (одношаговый)')
plt.xlabel('Часы (последние 168)')
plt.ylabel('Температура °C')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Сравнение метрик: Random Forest vs XGBoost
Сравнительная таблица метрик для RF, XGBoost (параметры Сочи) и XGBoost (своя оптимизация) для Санкт-Петербурга.

```python
from sklearn.metrics import mean_absolute_error


def calculate_wape(y_true, y_pred):
    return np.sum(np.abs(y_true - y_pred)) / np.sum(np.abs(y_true))

def calculate_smape(y_true, y_pred):
    y_true_arr = np.asarray(y_true)
    y_pred_arr = np.asarray(y_pred)
    denominator = np.abs(y_true_arr) + np.abs(y_pred_arr)
    return 100/len(y_true_arr) * np.sum(2 * np.abs(y_pred_arr - y_true_arr) / np.where(denominator == 0, 1, denominator))

def calculate_directional_accuracy(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    true_diff = np.sign(np.diff(y_true))
    pred_diff = np.sign(np.diff(y_pred))
    non_zero_true_diff_indices = np.where(true_diff != 0)
    if len(non_zero_true_diff_indices[0]) == 0:
        return np.nan
    correct_direction_count = np.sum(true_diff[non_zero_true_diff_indices] == pred_diff[non_zero_true_diff_indices])
    total_comparisons = len(non_zero_true_diff_indices[0])
    return correct_direction_count / total_comparisons

def calculate_directional_r2(y_true, y_pred):
    from sklearn.metrics import r2_score
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    true_diff = np.diff(y_true)
    pred_diff = np.diff(y_pred)
    if len(true_diff) < 2:
        return np.nan
    return r2_score(true_diff, pred_diff)

rf_params = {'n_estimators': 144, 'max_depth': 14, 'random_state': 42}

xgb_params_sochi = {
    'n_estimators': 304,
    'learning_rate': 0.04130706760619954,
    'max_depth': 6,
    'subsample': 0.8120878092460357,
    'colsample_bytree': 0.9488101466944081,
    'random_state': 42
}

model_rf_new = RandomForestRegressor(**rf_params, n_jobs=-1)
model_rf_new.fit(pd.concat([X_tr_new, X_val_new]), pd.concat([y_tr_new, y_val_new]))
preds_rf_new = model_rf_new.predict(X_te_new)

model_xgb_transfer = xgb.XGBRegressor(**xgb_params_sochi)
model_xgb_transfer.fit(pd.concat([X_tr_new, X_val_new]), pd.concat([y_tr_new, y_val_new]))
preds_xgb_transfer = model_xgb_transfer.predict(X_te_new)

mae_rf = mean_absolute_error(y_te_new, preds_rf_new)
smape_rf = calculate_smape(y_te_new, preds_rf_new)
wape_rf = calculate_wape(y_te_new, preds_rf_new)
da_rf = calculate_directional_accuracy(y_te_new, preds_rf_new)
dr2_rf = calculate_directional_r2(y_te_new, preds_rf_new)

mae_xgb_transfer = mean_absolute_error(y_te_new, preds_xgb_transfer)
smape_xgb_transfer = calculate_smape(y_te_new, preds_xgb_transfer)
wape_xgb_transfer = calculate_wape(y_te_new, preds_xgb_transfer)
da_xgb_transfer = calculate_directional_accuracy(y_te_new, preds_xgb_transfer)
dr2_xgb_transfer = calculate_directional_r2(y_te_new, preds_xgb_transfer)

mae_xgb_opt = mean_absolute_error(y_te_new, preds_xgb_opt)
smape_xgb_opt = calculate_smape(y_te_new, preds_xgb_opt)
wape_xgb_opt = calculate_wape(y_te_new, preds_xgb_opt)
da_xgb_opt = calculate_directional_accuracy(y_te_new, preds_xgb_opt)
dr2_xgb_opt = calculate_directional_r2(y_te_new, preds_xgb_opt)

metrics_comparison = pd.DataFrame({
    'Metric': [
        'MAE (°C)',
        'SMAPE (%)',
        'WAPE (%)',
        'Directional Accuracy',
        'Directional R²'
    ],
    'Random Forest': [
        f"{mae_rf:.4f}",
        f"{smape_rf:.2f}%",
        f"{wape_rf:.2%}",
        f"{da_rf:.4f}",
        f"{dr2_rf:.4f}"
    ],
    'XGBoost (параметры Сочи)': [
        f"{mae_xgb_transfer:.4f}",
        f"{smape_xgb_transfer:.2f}%",
        f"{wape_xgb_transfer:.2%}",
        f"{da_xgb_transfer:.4f}",
        f"{dr2_xgb_transfer:.4f}"
    ],
    'XGBoost (своя оптимизация)': [
        f"{mae_xgb_opt:.4f}",
        f"{smape_xgb_opt:.2f}%",
        f"{wape_xgb_opt:.2%}",
        f"{da_xgb_opt:.4f}",
        f"{dr2_xgb_opt:.4f}"
    ]
})

print(f"Сравнение моделей: {new_target_city} (одношаговый прогноз, тестовая выборка)")
display(metrics_comparison)

diff = mae_xgb_transfer - mae_xgb_opt
if abs(diff) < 0.03:
    print(f"Разница MAE = {diff:+.4f}°C, гиперпараметры отлично переносятся")
elif abs(diff) < 0.08:
    print(f"Разница MAE = {diff:+.4f}°C, гиперпараметры удовлитворительно переносятся")
else:
    print(f"Разница MAE = {diff:+.4f}°C, для каждого города лучше подбирать свои гиперпараметры")
```

### Рекурсивная стратегия прогнозирования на 168 часов (1 неделя)
В отличие от точечного прогноза, рекурсивный метод использует собственные предсказания модели для формирования лагов на будущие шаги, из-за чего растет ошибка

```python
def recursive_forecast(model, df_initial, features, steps=168):

    forecast = []
    current_data = df_initial.iloc[0:1].copy()
    buffer = []

    for lag in [1, 2, 3, 6, 12, 24, 48, 168]:
        col = f'lag_{lag}'
        if col in current_data.columns:
            buffer.append(current_data[col].iloc[0])
        else:
            buffer.append(np.nan)
    while len(buffer) < 168:
        buffer.insert(0, np.nan)

    for i in range(steps):
        pred = model.predict(current_data[features])[0]
        forecast.append(pred)
        buffer.append(pred)

        new_ds = current_data['ds'].iloc[0] + pd.Timedelta(hours=1)

        new_row = {
            'ds': new_ds,
            'hour': new_ds.hour,
            'dayofweek': new_ds.dayofweek,
            'month': new_ds.month,
            'dayofyear': new_ds.dayofyear,
            'is_weekend': 1 if new_ds.dayofweek in [5, 6] else 0,
            'hour_sin': np.sin(2 * np.pi * new_ds.hour / 24),
            'hour_cos': np.cos(2 * np.pi * new_ds.hour / 24),
            'month_sin': np.sin(2 * np.pi * new_ds.month / 12),
            'month_cos': np.cos(2 * np.pi * new_ds.month / 12),
        }

        for lag in [1, 2, 3, 6, 12, 24, 48, 168]:
            if len(buffer) >= lag:
                new_row[f'lag_{lag}'] = buffer[-lag]
            else:
                new_row[f'lag_{lag}'] = current_data[f'lag_{lag}'].iloc[0]

        for window in [6, 12, 24]:
            if len(forecast) >= window:
                recent = forecast[-window:]
                new_row[f'rolling_mean_{window}'] = np.mean(recent)
                new_row[f'rolling_std_{window}'] = np.std(recent) if len(recent) > 1 else 0
            else:
                new_row[f'rolling_mean_{window}'] = current_data[f'rolling_mean_{window}'].iloc[0]
                new_row[f'rolling_std_{window}'] = current_data[f'rolling_std_{window}'].iloc[0]

        current_data = pd.DataFrame([new_row])

    return forecast

test_start_idx = val_idx
initial_state = X_test.iloc[0:1].copy()
initial_state['ds'] = city_df['ds'].iloc[test_start_idx]

horizon = 168
recursive_preds = recursive_forecast(best_xgb, initial_state, features, steps=horizon)

plt.figure(figsize=(15, 6))
plt.plot(range(horizon), y_test.values[:horizon],
         label='Реальность', color='black', marker='.')
plt.plot(range(horizon), recursive_preds,
         label='Рекурсивный прогноз (7 дней)', color='red', linestyle='--')
plt.title(f'Рекурсивный прогноз на 168 часов: {target_city}')
plt.xlabel('Часы вперед')
plt.ylabel('Температура °C')
plt.legend()
plt.grid(True)
plt.show()

rec_mae = mean_absolute_error(y_test.values[:horizon], recursive_preds)
rec_smape = calculate_smape(y_test.values[:horizon], recursive_preds)
rec_wape = calculate_wape(y_test.values[:horizon], recursive_preds)
rec_da = calculate_directional_accuracy(y_test.values[:horizon], recursive_preds)

print(f"Рекурсивный прогноз на 168 часов:")
print(f"MAE:  {rec_mae:.4f}")
print(f"SMAPE: {rec_smape:.2f}%")
print(f"WAPE:  {rec_wape:.4%}")
print(f"Dir.Acc: {rec_da:.4f}")
```
